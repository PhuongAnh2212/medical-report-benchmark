```python
# Optional: simple bar chart comparing models on BLEU-4
import matplotlib.pyplot as plt

if not leaderboard.empty and 'BLEU-4' in leaderboard.columns:
    plt.figure(figsize=(6, 4))
    plt.bar(leaderboard['Model'], leaderboard['BLEU-4'])
    plt.ylabel('BLEU-4')
    plt.title('Model comparison: BLEU-4')
    plt.xticks(rotation=30, ha='right')
    plt.tight_layout()
    plt.show()
```
